### Simple Linear Regression

```python
# create an environment and then install the requirements in that environment
# then, run this notebook in that environment/kernel

import os
import sys

os.makedirs("figures", exist_ok=True)
sys.path.append(os.path.abspath(".."))
```

```python
from matplotlib import pyplot as plt

SEED = 42
```

```python
from src import linear_regression

y_test, y_pred = linear_regression(SEED)
plt.figure()
plt.hist(y_test - y_pred, bins=40)
plt.title("Residuals (y - y_pred): Linear Regression")
plt.xlabel("Residual")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig(os.path.join("figures", "residuals_lin_reg.png"))
plt.show()

plt.figure()
plt.scatter(y_test, y_pred, s=6)
plt.title("Predicted vs True: Linear Regression")
plt.xlabel("True")
plt.ylabel("Pred")
plt.tight_layout()
plt.savefig(os.path.join("figures", "pred_vs_true_lin_reg.png"))
plt.show()
```
